# Thống kê thực hành và A/B test

## Mục tiêu
- Phân biệt population, sample, statistic và uncertainty.
- Dùng bootstrap để ước lượng confidence interval.
- Đọc A/B test bằng effect size, CI và p-value.

```python
import numpy as np
from scipy.stats import ttest_ind

rng = np.random.default_rng(42)
control = rng.normal(100, 20, size=400)
treatment = rng.normal(106, 20, size=400)
observed_lift = treatment.mean() - control.mean()
observed_lift
```

## Bootstrap confidence interval
Bootstrap lấy mẫu có hoàn lại từ sample hiện có. Nó ước lượng độ không chắc chắn của statistic; không sửa được sample bias hay thiết kế thí nghiệm sai.

```python
bootstrap_lifts = np.array([
    rng.choice(treatment, size=len(treatment), replace=True).mean() -
    rng.choice(control, size=len(control), replace=True).mean()
    for _ in range(2_000)
])
ci_95 = np.percentile(bootstrap_lifts, [2.5, 97.5])
ci_95
```

```python
test = ttest_ind(treatment, control, equal_var=False)
{'lift': observed_lift, 'p_value': test.pvalue, 'ci_95': ci_95}
```

## Diễn giải an toàn
- `p-value` không phải xác suất giả thuyết null đúng.
- CI cho khoảng giá trị effect tương thích với data và giả định.
- Quyết định còn cần MDE, sample-ratio mismatch, guardrail metric, thời gian chạy và chi phí.
- Không dừng thử nghiệm khi mới thấy kết quả thuận lợi (*peeking*). Xác định sample size và rule trước khi chạy.
