## Mục tiêu
- Đọc file qua `pathlib` thay vì phụ thuộc thư mục đang mở.
- Phân biệt lỗi dự kiến và lỗi cần sửa.
- Dùng `dataclass`, generator và type hint khi chúng giúp code dễ đọc.

```python
from pathlib import Path
import csv

repo_root = Path('.') if Path('data').exists() else Path('..')
data_path = repo_root / 'data/retail_customers.csv'
with data_path.open(encoding='utf-8', newline='') as file:
    rows = list(csv.DictReader(file))
rows[0]
```

## Xử lý lỗi đúng chỗ
Chỉ bắt exception mà bạn có kế hoạch xử lý. Không dùng `except Exception: pass`, vì nó che giấu bug và tạo kết quả sai yên lặng.

```python
def parse_income(value: str) -> int | None:
    if value in ('', 'not_available'):
        return None
    try:
        return int(value)
    except ValueError as error:
        raise ValueError(f'income không hợp lệ: {value!r}') from error

assert parse_income('18000000') == 18_000_000
assert parse_income('not_available') is None
```

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Metric:
    name: str
    value: float
    unit: str = ''

Metric('conversion_rate', 0.082, '%')
```
