# Unsupervised learning: clustering và PCA

## Mục tiêu
- Chuẩn hoá feature trước các thuật toán dựa vào khoảng cách.
- Dùng KMeans để tạo hypothesis, không gắn nhãn khách hàng một cách tuyệt đối.
- Dùng PCA để quan sát/giảm chiều và hiểu trade-off variance.

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.4, random_state=42)
X_scaled = StandardScaler().fit_transform(X)
```

```python
scores = {}
for k in range(2, 7):
    labels = KMeans(n_clusters=k, n_init='auto', random_state=42).fit_predict(X_scaled)
    scores[k] = silhouette_score(X_scaled, labels)
scores
```

```python
model = KMeans(n_clusters=3, n_init='auto', random_state=42).fit(X_scaled)
projection = PCA(n_components=2, random_state=42).fit_transform(X_scaled)
plt.scatter(projection[:, 0], projection[:, 1], c=model.labels_, cmap='viridis', alpha=.7)
plt.title('Clusters projected to two PCA components')
```

## Thận trọng khi segmentation
Silhouette cao không chứng minh segment có ý nghĩa business, bền theo thời gian hay công bằng. Đọc profile từng cluster, kiểm tra stability theo sample và đặt tên segment theo hành vi quan sát được, không theo định kiến.
