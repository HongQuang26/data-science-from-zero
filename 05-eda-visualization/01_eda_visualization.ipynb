# EDA và trực quan hoá có mục đích

## Mục tiêu
- Biến câu hỏi business thành phép kiểm tra với data.
- Chọn chart theo kiểu biến.
- Viết insight có bằng chứng và giới hạn.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

repo_root = Path('.') if Path('data').exists() else Path('..')
df = pd.read_csv(repo_root / 'data/retail_customers.csv')
df['income'] = pd.to_numeric(df['income'], errors='coerce')
sns.set_theme(style='whitegrid')
```

## Câu hỏi: thành phố nào có average spend cao hơn?
Bar chart phù hợp để so sánh aggregate theo category. Bar không thể hiện phân phối từng khách hàng; khi cần, dùng boxplot/stripplot bổ sung.

```python
summary = df.groupby('city', as_index=False)['spend'].mean().sort_values('spend', ascending=False)
ax = sns.barplot(data=summary, x='city', y='spend', hue='city', legend=False)
ax.set(title='Average spend by city', xlabel='', ylabel='VND')
ax.tick_params(axis='x', rotation=20)
plt.tight_layout()
```

```python
# Relationship: số lần ghé thăm và spend.
ax = sns.scatterplot(data=df, x='visits', y='spend', hue='channel', s=80)
ax.set(title='Visits vs. spend', xlabel='Visits', ylabel='Spend (VND)')
plt.tight_layout()
```

## Cách viết insight
Ví dụ: *Trong mẫu rất nhỏ này, Đà Nẵng có mean spend cao nhất. Đây là tín hiệu để kiểm tra tiếp, không phải lý do đủ để chuyển ngân sách: thành phố có ít quan sát và một khách spend cao có thể kéo mean.*

Tách rõ **fact trong data**, **diễn giải**, **hành động đề xuất** và **limitation**.
