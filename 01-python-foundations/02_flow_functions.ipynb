# Python: flow control và functions

## Mục tiêu
- Dùng `if`, `for`, `while` có chủ đích.
- Đóng gói logic lặp lại trong function có tên rõ ràng.
- Xử lý input rỗng trước khi tính toán.

```python
spend = 1_250_000
if spend >= 1_000_000:
    tier = 'gold'
elif spend >= 500_000:
    tier = 'silver'
else:
    tier = 'new'
tier
```

```python
orders = [250_000, 380_000, 175_000]
total = 0
for amount in orders:
    total += amount
total

# Comprehension rõ khi chỉ biến đổi/lọc một lần.
high_value = [amount for amount in orders if amount >= 250_000]
high_value
```

## Function tốt
Function nhận input, trả output và làm một việc. Không `print` thay cho `return`: `return` giúp test và tái sử dụng kết quả.

```python
def average(values):
    """Trả về trung bình; trả None khi không có quan sát."""
    if not values:
        return None
    return sum(values) / len(values)

assert average([2, 4, 6]) == 4
assert average([]) is None
average(orders)
```
