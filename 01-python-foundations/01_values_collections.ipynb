## Collections thường dùng
- `list`: có thứ tự, thay đổi được; dùng cho một dãy quan sát.
- `tuple`: có thứ tự, không thay đổi; dùng cho record cố định.
- `dict`: ánh xạ key → value; dùng cho một record có tên cột.
- `set`: phần tử duy nhất; dùng để tìm category trùng.

```python
scores = [8.5, 7.0, 9.0, 7.0]
coordinate = (10.78, 106.70)
student = {'name': 'An', 'track': 'DS', 'score': 8.5}
unique_scores = set(scores)
scores[-1], student['name'], unique_scores
```
