# Pandas: bảng dữ liệu có ý nghĩa

## Mục tiêu
- Đọc CSV và audit nhanh DataFrame.
- Lọc, group, aggregate và reshape dữ liệu.
- Tránh gán vào slice mơ hồ.

```python
from pathlib import Path
import pandas as pd

repo_root = Path('.') if Path('data').exists() else Path('..')
df = pd.read_csv(repo_root / 'data/retail_customers.csv')
df.head(), df.shape
```

```python
# Audit tối thiểu trước khi phân tích.
df.info()
df.isna().sum()
df.nunique()
```

```python
# Chuyển kiểu có kiểm soát và tạo subset với .loc.
df['income'] = pd.to_numeric(df['income'], errors='coerce')
recent_online = df.loc[(df['channel'] == 'online') & (df['last_purchase_days'] <= 30)].copy()
city_summary = (df.groupby('city', dropna=False)
                .agg(customers=('customer_id', 'count'), spend=('spend', 'mean'))
                .sort_values('spend', ascending=False))
city_summary
```
