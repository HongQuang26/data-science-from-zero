# NumPy: tính toán vector hoá

## Mục tiêu
- Hiểu `ndarray`, `shape`, `dtype` và broadcasting.
- Thay loop Python chậm bằng thao tác mảng khi phù hợp.
- Kiểm tra shape trước phép toán.

```python
import numpy as np
sales = np.array([1.2, 2.5, 0.8, 3.1]) * 1_000_000
sales, sales.shape, sales.dtype
```

```python
growth_rate = 0.10
next_month_plan = sales * (1 + growth_rate)
mean_sales = sales.mean()
above_average = sales[sales > mean_sales]
next_month_plan, above_average
```

## Broadcasting
NumPy so khớp dimension từ phải sang trái. Broadcasting tiện nhưng có thể tạo kết quả sai nếu `shape` không đúng, vì vậy luôn in shape khi biến đổi mảng 2D.

```python
matrix = np.array([[10, 20, 30], [15, 25, 35]])
channel_target = np.array([12, 22, 32])
gap = matrix - channel_target
assert gap.shape == matrix.shape
gap
```
